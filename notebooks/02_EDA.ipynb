```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(
    "../data/raw/column_2C.dat",
    sep=r"\s+",
    header=None
)

df.columns = [
    "pelvic_incidence",
    "pelvic_tilt",
    "lumbar_lordosis_angle",
    "sacral_slope",
    "pelvic_radius",
    "degree_spondylolisthesis",
    "class"
]
```

```python
sns.countplot(data=df, x="class")

plt.title("Distribution of Normal and Abnormal Classes")
plt.xlabel("Spinal Condition")
plt.ylabel("Number of Records")
plt.show()
```

```python
df.hist()
plt.show()
```

```python
sns.boxplot(data=df, x="class", y="pelvic_incidence")

plt.title("Pelvic Incidence by Spinal Condition")
plt.xlabel("Spinal Condition")
plt.ylabel("Pelvic Incidence")
plt.show()
```

```python
sns.boxplot(data=df, x="class", y="pelvic_tilt")

plt.title("Pelvic Tilt by Spinal Condition")
plt.xlabel("Spinal Condition")
plt.ylabel("Pelvic Tilt")
plt.show()
```

```python
sns.boxplot(data=df, x="class", y="lumbar_lordosis_angle")

plt.title("Lumbar Lordosis Angle by Spinal Condition")
plt.xlabel("Spinal Condition")
plt.ylabel("Lumbar Lordosis Angle")
plt.show()
```

```python
sns.boxplot(data=df, x="class", y="sacral_slope")

plt.title("Sacral Slope by Spinal Condition")
plt.xlabel("Spinal Condition")
plt.ylabel("Sacral Slope")
plt.show()
```

```python
sns.boxplot(data=df, x="class", y="pelvic_radius")

plt.title("Pelvic Radius by Spinal Condition")
plt.xlabel("Spinal Condition")
plt.ylabel("Pelvic Radius")
plt.show()
```

```python
sns.boxplot(data=df, x="class", y="degree_spondylolisthesis")

plt.title("Degree of Spondylolisthesis by Spinal Condition")
plt.xlabel("Spinal Condition")
plt.ylabel("Degree of Spondylolisthesis")
plt.show()
```

```python
correlation = df.corr(numeric_only=True)

sns.heatmap(correlation)

plt.show()
```

```python
sns.scatterplot(data=df, x="pelvic_incidence", y="pelvic_tilt", hue="class")

plt.title("Pelvic Incidence vs Pelvic Tilt")
plt.xlabel("Pelvic Incidence")
plt.ylabel("Pelvic Tilt")
plt.show()
```

```python
sns.scatterplot(data=df, x="lumbar_lordosis_angle", y="sacral_slope", hue="class")

plt.title("Lumbar Lordosis Angle vs Sacral Slope")
plt.xlabel("Lumbar Lordosis Angle")
plt.ylabel("Sacral Slope")
plt.show()
```
