```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(
    "../data/raw/column_2C.dat",
    sep=r"\s+",
    header=None
)

df.columns = [
    "pelvic_incidence",
    "pelvic_tilt",
    "lumbar_lordosis_angle",
    "sacral_slope",
    "pelvic_radius",
    "degree_spondylolisthesis",
    "class"
]
```

```python
sns.countplot(data=df, x="class")

plt.title("Distribution of Normal and Abnormal Classes")
plt.xlabel("Spinal Condition")
plt.ylabel("Number of Records")
plt.show()
```

```python
df.hist()
plt.show()
```

```python
sns.boxplot(data=df, x="class", y="pelvic_incidence")

plt.title("Pelvic Incidence by Spinal Condition")
plt.xlabel("Spinal Condition")
plt.ylabel("Pelvic Incidence")
plt.show()
```
